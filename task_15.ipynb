```python
import pandas as pd
from sklearn.datasets import load_iris

iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df['species'] = [
    iris.target_names[i] for i in iris.target
]

print(df.head())
```

```python
print("Shape:", df.shape)

print("\nColumns:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nInformation:")
df.info()

print("\nSummary Statistics:")
print(df.describe())
```

```python
print("First 10 rows:")
print(df.head(10))

print("\nSepal Length:")
print(df['sepal length (cm)'])

print("\nSelected Columns:")
print(df[['sepal length (cm)', 'petal length (cm)', 'species']])
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
print("Species Count:")
print(df['species'].value_counts())
```
